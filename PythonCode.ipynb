 Prime
number

```python
n = int(input("enter a number"))
count=0
for i in range(1, n + 1):
    if n%i==0
        count++

    if count==2{
        print("prime number")
    }
    else{
        print("not a prime number")
    }
}
```

factorial programs

```python
n= int(input("enter the value of n"))
def factorial(n):
  if n==0:
    return 1
  else:
    c= n*factorial(n-1)
    return c
print("factorial is", factorial(n))
```

Fibonacci series using recursion

```python
n= int(input("enter the value of n"))
def fibonacci(n):

  if n==0:
    return 0
  elif n==1:
    return 1
  else:
    return fibonacci(n-1)+fibonacci(n-2)

print("fibonacci series is",fibonacci(n))
```

```python
n = int(input("Enter n: "))
fibo = lambda n: n if n<=1 else fibo(n-1)+fibo(n-2)
if__name__=="__main__":
print("Fibonacci series is", fibo(n))
```
